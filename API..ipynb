# `Automatic Image downloading in folder`

```python
def Image_search(search_name):
    import requests
    import os
    url = "https://real-time-image-search.p.rapidapi.com/search"

    querystring = {"query":search_name,"limit":"10","size":"any","color":"any","type":"any","time":"any","usage_rights":"any","file_type":"any","aspect_ratio":"any","safe_search":"off","region":"us"}
    
    headers = {
    	"x-rapidapi-key": "7cbfcfe4ecmshc8885e0effbd816p145679jsnc7988faadadc",
    	"x-rapidapi-host": "real-time-image-search.p.rapidapi.com",
    	"Content-Type": "application/json"
    }
    
    response = requests.get(url, headers=headers, params=querystring)
    if response.status_code == 200:
        data = response.json()
        url_list =[]
        for i in range(0,len(data["data"])):
            url_list.append(data['data'][i]['thumbnail_url'])
    
        current_path = os.getcwd()
        os.mkdir(search_name)
        os.chdir(search_name)
        print(f"Enter into {os.getcwd()}")
    
        c=1
        for url in url_list:
            resp = requests.get(url)
            img =resp.content
            with open(f"img{c}.jpeg","wb") as file:
                file.write(img)
                c+=1
        else:
            print("Image downloaded...")
            os.chdir(current_path)
            print(f"finally in {os.getcwd()}")
        
    
    else:
        print(response.reason)

```

```python
Image_search("cars")
```

# `Take data from rapid api and fetch data and show in pandas`

```python
import requests

url = "https://jsearch.p.rapidapi.com/search"

querystring = {"query":"developer jobs in chicago","page":"1","num_pages":"2","country":"us","date_posted":"all"}

headers = {
	"x-rapidapi-key": "7cbfcfe4ecmshc8885e0effbd816p145679jsnc7988faadadc",
	"x-rapidapi-host": "jsearch.p.rapidapi.com",
	"Content-Type": "application/json"
}

response = requests.get(url, headers=headers, params=querystring)

data = response.json()
```

# `collect details`

```python
title = []
emp_name =[]
job_publish = []
job_employment_type = []
job_apply_link = []
job_posted_at =[]
job_city = []
job_country = []
for i in range(len(data['data'])):
    title.append(data['data'][i]['job_title'])
    emp_name.append(data['data'][i]['employer_name'])
    job_publish.append(data['data'][i]['job_publisher'])
    job_employment_type.append(data['data'][i]['job_employment_type'])
    job_apply_link.append(data['data'][i]['job_apply_link'])
    job_posted_at.append(data['data'][i] ['job_posted_at'])
    job_city.append(data['data'][i] ['job_city'])
    job_country.append(data['data'][i]['job_country'])
else:
    print("Done")
```

# `now show in pandas`

```python
import pandas as pd 
data = pd.DataFrame({
    "job title" : title,
    "Employer_name": emp_name,
    "job_pulisher" :job_publish,
    "job_employment_type": job_employment_type,
    "job_apply_link":job_apply_link,
    "job_posted_at":job_posted_at,
    "job city": job_city,
    "job country": job_country
})
data
```

# `now convert in csv file`

```python
data.to_csv("linked.csv")
```
