# Validate Gmail id
- gmail ends with @gmail.com
- cannot starts with digit
- can not contail any symbol in mid or starting (@ , .)

```python
mail = input("Enter your valid gmail id")
if mail.endswith("@gmail.com") and not mail[0].isdigit() and mail.replace("@gmail.com" , "").isalnum():
    print("Valid mail")

else:
    print("Invalid mail")
```
