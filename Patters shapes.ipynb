# Right angle print

```python
i =1
while i <= 5: # no. of row 5
    j=1
    while j<=5: # no. of col. 5
        if j<=i:  # here >>> condition
            print("*", end=" ")
        else:
            print(" ",end=" ")
        j+=1
    print()    
    i+=1
```

# opposite of right angle

```python
i =1
while i <= 5:
    j=1
    while j<=5:
        if j>=i:
            print("*", end=" ")
        else:
            print(" ",end=" ")
        j+=1
    print()    
    i+=1
```

# Make the square by printing stars by taking input 

```python
num = int(input("enter the number: "))
i =1
while i <= num:
    j=1
    while j<=num:
        if j>=i or j<=i:
            print("*", end=" ")
        else:
            print(" ",end=" ")
        j+=1
    print()    
    i+=1
```

# Hollow square

```python
 
i =1
while i <= 5:
    j=1
    while j<=5:
        if i==1 or i==5 or j==1 or j==5:
            print("*", end=" ")
        else:
            print(" ",end=" ")
        j+=1
    print()    
    i+=1
```

# Inverted triangle

```python
n = 5
i = n
while i > 0:
    print("*" * i)
    i -= 1
```

# Pyramid pattern

```python
n = 5
i = 1
while i <= n:
    spaces = n - i
    stars = 2 * i - 1
    print(" " * spaces + "*" * stars)
    i += 1
```

# Diamond Pattern

```python
n = 5

# upper half
i = 1
while i <= n:
    print(" " * (n - i) + "*" * (2 * i - 1))
    i += 1

# lower half
i = n - 1
while i > 0:
    print(" " * (n - i) + "*" * (2 * i - 1))
    i -= 1
```

# Number patter

```python
n = 5
i = 1
while i <= n:
    j = 1
    while j <= i:
        print(j, end="")
        j += 1
    print()
    i += 1
```
