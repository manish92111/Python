# (A TO Z) PATTERN PRINTING

```python
n = 5

def print_letter(ch):
    for i in range(n):
        for j in range(n):

            if ch == 'A':
                if i == 0 or i == n//2 or j == 0 or j == n-1:
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'B':
                if j == 0 or i == 0 or i == n//2 or i == n-1 or j == n-1:
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'C':
                if i == 0 or i == n-1 or j == 0:
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'D':
                if j == 0 or i == 0 or i == n-1 or j == n-1:
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'E':
                if i == 0 or i == n//2 or i == n-1 or j == 0:
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'F':
                if i == 0 or i == n//2 or j == 0:
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'G':
                if i == 0 or i == n-1 or j == 0 or (i >= n//2 and j == n-1):
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'H':
                if j == 0 or j == n-1 or i == n//2:
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'I':
                if i == 0 or i == n-1 or j == n//2:
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'J':
                if i == 0 or j == n//2 or (i == n-1 and j <= n//2):
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'K':
                if j == 0 or i+j == n//2 or i-j == n//2:
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'L':
                if j == 0 or i == n-1:
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'M':
                if j == 0 or j == n-1 or i == j and i <= n//2 or i+j == n-1 and i <= n//2:
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'N':
                if j == 0 or j == n-1 or i == j:
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'O':
                if i == 0 or i == n-1 or j == 0 or j == n-1:
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'P':
                if j == 0 or i == 0 or i == n//2 or (j == n-1 and i <= n//2):
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'Q':
                if i == 0 or j == 0 or i == n-2 or j == n-2 or (i == j and i >= n//2):
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'R':
                if j == 0 or i == 0 or i == n//2 or (j == n-1 and i <= n//2) or i-j == n//2:
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'S':
                if i == 0 or i == n//2 or i == n-1 or (j == 0 and i < n//2) or (j == n-1 and i > n//2):
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'T':
                if i == 0 or j == n//2:
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'U':
                if j == 0 or j == n-1 or i == n-1:
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'V':
                if (j == 0 and i < n-1) or (j == n-1 and i < n-1) or (i == n-1 and j == n//2):
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'W':
                if j == 0 or j == n-1 or (i == j and i >= n//2) or (i+j == n-1 and i >= n//2):
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'X':
                if i == j or i+j == n-1:
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'Y':
                if (i == j and i <= n//2) or (i+j == n-1 and i <= n//2) or (j == n//2 and i > n//2):
                    print("*", end="")
                else: print(" ", end="")

            elif ch == 'Z':
                if i == 0 or i == n-1 or i+j == n-1:
                    print("*", end="")
                else: print(" ", end="")

        print()


# Print A to Z
for ch in range(ord('A'), ord('Z') + 1):
    print("\n" + chr(ch))
    print_letter(chr(ch))
```
